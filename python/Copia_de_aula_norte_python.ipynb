```python
def limpiar_email(email):
    return email.strip().lower()

def limpiar_nombre(nombre):
    return nombre.strip().title()

def limpiar_telefono(telefono):
    digitos = ""
    for c in telefono:
        if c.isdigit():
            digitos = digitos + c
    if digitos == "":
        return None
    return digitos

def limpiar_interesado(interesado):
    return {
        "nombre": limpiar_nombre(interesado["nombre"]),
        "email": limpiar_email(interesado["email"]),
        "telefono": limpiar_telefono(interesado["telefono"]),
    }

import requests
from google.colab import userdata

url = "TU_TEST_URL"
clave = userdata.get("N8N_CLAVE")

sucios = [
    {"nombre": "  lAURA ", "email": " Laura@Mail.com", "telefono": "600 11 22 33",
     "curso": "Excel avanzado", "mensaje": "¿Cuánto cuesta?"},
    {"nombre": "MARCOS", "email": "marcos@MAIL.com ", "telefono": "+34 611-22-33-44",
     "curso": "Marketing digital", "mensaje": "quiero empezar ya"},
    {"nombre": "sofía", "email": "sofia@mail.com", "telefono": "no tengo",
     "curso": "Inglés para entrevistas", "mensaje": "más adelante quizás"},
    {"nombre": " juan pablo", "email": "JP@mail.com", "telefono": "622334455",
     "curso": "Cocina", "mensaje": "hola"},
    {"nombre": "Elena", "email": "elena-sin-arroba.com", "telefono": "633445566",
     "curso": "Excel avanzado", "mensaje": "info"},
]

for s in sucios:
    limpio = limpiar_interesado(s)
    limpio["curso"] = s["curso"]
    limpio["mensaje"] = s["mensaje"]
    r = requests.post(url, json=limpio, headers={"X-Clave-Aula": clave})
    print(limpio["email"], "→", r.status_code)
```
